```python
import numpy as np

def mamba(x,d_state):
  seq_len,d_model=x.shape
  A=np.random.randn(d_state,)
  B=np.random.randn(d_model,d_state)
  C=np.random.randn(d_state,d_model)
  delta=np.random.randn(d_model,d_state)

  delta=np.log(1+np.exp(x@delta))
  A_bar=np.exp(delta*A)
  B_bar=delta*(x@B)

  h=np.zeros(d_state)
  outputs=[]
  for i in range(seq_len):
    h=A_bar[i]*h+B_bar[i]
    y_t=h@C
    outputs.append(y_t)

  output=np.array(outputs)
  output=output/(1+np.exp(-1*output))
  return output
```

```python
# test
np.random.seed(42)
seq_len=4
d_model=16
d_state=8

x=np.random.randn(seq_len,d_model)
output=mamba(x,d_state)
print('output shape: ',output.shape)      # (4,16)
```
