```python
import numpy as np

def mamba_block(x,d_state):
  seq_len,d_model=x.shape     # x: (seq_len,d_model)

  # A,B,C,delta 초기화
  A=np.random.randn(d_state,)     # 각 상태마다 하나
  B=np.random.randn(d_model,d_state)      # 입력 -> 상태
  C=np.random.randn(d_state,d_model)      # 상태 -> 출력
  delta=np.random.randn(d_model,d_state)      # B랑 같은 크기

  # delta 계산(softplus)
  delta=np.log(1+np.exp(x@delta))

  # A_bar,B_bar 이산화
  A_bar=np.exp(delta*A)
  B_bar=delta*(x@B)

  # SSM 루프(h_t=A_bar*h+B_bar*x)
  h=np.zeros(d_state)
  y=[]
  for i in range(seq_len):
    h=A_bar[i]*h+B_bar[i]
    y_t=h@C
    y.append(y_t)
  output=np.array(y)
  # Gating(SiLU)
  output=np.maximum(0,output)*(1/(1+np.exp(-1*output)))

  return output
```

```python
# test
np.random.seed(42)
seq_len=4
d_model=16
d_state=8

x=np.random.randn(seq_len,d_model)
output=mamba_block(x,d_state)
print('output shape: ',np.array(output).shape)      # (4,16)
```
